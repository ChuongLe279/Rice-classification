```python
from pathlib import Path
import sys

import cv2
import matplotlib.pyplot as plt
import numpy as np

# Tìm root khi mở notebook từ root, src hoặc src/alexnet.
PROJECT_ROOT = next(
    (p for p in (Path.cwd(), *Path.cwd().parents)
     if (p / 'src' / 'utils.py').is_file() and (p / 'Rice dataset').is_dir()),
    None,
)
if PROJECT_ROOT is None:
    raise FileNotFoundError('Hãy chạy notebook từ bên trong thư mục CV-Rice.')
SRC_DIR = str(PROJECT_ROOT / 'src')
if SRC_DIR not in sys.path:
    sys.path.insert(0, SRC_DIR)

from utils import load_images_from_folder
```

## 1. Đọc dữ liệu

```python
DATASET_PATH = PROJECT_ROOT / 'Rice dataset' / 'Rice_Image_Dataset'
images, labels, class_names = load_images_from_folder(str(DATASET_PATH))
labels = np.asarray(labels, dtype=np.int64)
if not images:
    raise ValueError(f'Không đọc được ảnh nào trong {DATASET_PATH}')

class_counts = np.bincount(labels, minlength=len(class_names))
print(f'Tổng cộng: {len(images):,} ảnh | {len(class_names)} class')
fig, ax = plt.subplots(figsize=(7, 3.5))
bars = ax.bar(class_names, class_counts, color='steelblue', width=0.6)
ax.bar_label(bars, labels=[f'{count:,}' for count in class_counts], padding=3)
ax.set(title='Số lượng ảnh theo từng class', xlabel='Class', ylabel='Số ảnh')
ax.set_ylim(0, class_counts.max() * 1.15)
ax.spines[['top', 'right']].set_visible(False)
fig.tight_layout()
plt.show()
```

```python
# Mỗi hàng là một class, lấy ngẫu nhiên tối đa 3 ảnh để minh họa.
SAMPLES_PER_CLASS = 3
rng = np.random.default_rng(42)
fig, axes = plt.subplots(
    len(class_names), SAMPLES_PER_CLASS,
    figsize=(5, 1.35 * len(class_names)), squeeze=False,
)
for class_id, class_name in enumerate(class_names):
    indices = np.flatnonzero(labels == class_id)
    selected = rng.choice(indices, size=min(SAMPLES_PER_CLASS, len(indices)), replace=False)
    for column, ax in enumerate(axes[class_id]):
        ax.set_xticks([])
        ax.set_yticks([])
        for spine in ax.spines.values():
            spine.set_visible(False)
        if column < len(selected):
            ax.imshow(cv2.cvtColor(images[selected[column]], cv2.COLOR_BGR2RGB))
        if column == 0:
            ax.set_ylabel(class_name, rotation=0, ha='right', va='center', labelpad=8)
fig.suptitle('Ảnh minh họa từng class', fontsize=12)
fig.tight_layout(rect=(0, 0, 1, 0.96), pad=0.5)
plt.show()
```
